```python
import pandas as pd
import json
```

```python
# Step 1: Create sample JSON data
data = [
    {"id": 1, "name": "Laptop", "price": 800},
    {"id": 2, "name": "Mouse", "price": 20},
    {"id": 3, "name": "Keyboard", "price": 50}
]
```

```python
# Step 2: Save JSON to a file
with open('data.json', 'w') as f:
    json.dump(data, f)
```

```python
# Step 3: Load JSON into Pandas DataFrame
df_json = pd.read_json('data.json')
```

```python

# Step 4: Display DataFrame
print("JSON Data:")
print(df_json)
```
