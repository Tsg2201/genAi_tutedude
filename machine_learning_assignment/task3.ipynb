```python
import sqlite3
import pandas as pd
```

```python
# Step 1: Create SQLite database
conn = sqlite3.connect('sample.db')
cursor = conn.cursor()
```

```python
# Step 2: Create employees table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL
)
""")
```

```python
# Step 3: Insert at least 5 records
employees_data = [
    (1, 'Alice', 'HR', 50000),
    (2, 'Bob', 'IT', 70000),
    (3, 'Charlie', 'Finance', 65000),
    (4, 'David', 'IT', 72000),
    (5, 'Eva', 'Marketing', 60000)
]

cursor.executemany(
    "INSERT OR REPLACE INTO employees VALUES (?, ?, ?, ?)",
    employees_data
)

conn.commit()
```

```python
# Step 4: Read table into Pandas DataFrame
df_sql = pd.read_sql_query("SELECT * FROM employees", conn)
```

```python
# Step 5: Display DataFrame
print("\nSQL Data:")
print(df_sql)
```

```python

# Close connection
conn.close()
```
